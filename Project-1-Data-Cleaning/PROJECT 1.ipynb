```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_excel(r"C:\Users\Hamza Ali\Downloads\Dataset for Data Analytics.xlsx")

print("===== DATA CLEANING PROJECT =====")
```

```python
# CHECK MISSING VALUES
print("\nMissing Values Before Cleaning:")
print(data.isnull().sum())
```

```python
# HANDLE MISSING VALUES
data["CouponCode"] = data["CouponCode"].fillna("No Coupon")
```

```python
# CHECK DUPLICATE RECORDS
duplicates = data.duplicated().sum()

print("\nNumber of Duplicate Rows:", duplicates)
```

```python
# REMOVE DUPLICATE ROWS
data = data.drop_duplicates()
```

```python
# CHECK DUPLICATE ORDE IDs
duplicate_order_ids = data["OrderID"].duplicated().sum()

print("\nNumber of Duplicate Order IDs:",
duplicate_order_ids)
```

```python
# VALIDATE DATE FORMAT
data["Date"] = pd.to_datetime(data["Date"])

invalid_dates = data["Date"].isnull().sum()
```

```python
# STANDARDIZE TEXT FORMATTING
data["Product"] = data["Product"].str.title()

data["PaymentMethod"] = data["PaymentMethod"].str.title()

data["OrderStatus"] = data["OrderStatus"].str.title()

data["ReferralSource"] = data["ReferralSource"].str.title()
```

```python
# FINAL VERIFICATION
print("\n===== FINAL REPORT =====")

print("\nMissing Values After Cleaning:")
print(data.isnull().sum())

print("\nDuplicate Rows Remaining:",
data.duplicated().sum())

print("Duplicate Order IDs Remaining:",
data["OrderID"].duplicated().sum())

print("Incorrectly Formatted Dates:",
invalid_dates)
```

```python
# SAVE CLEANED DATASET
data.to_excel(
r"C:\Users\Hamza Ali\Downloads\Cleaned_Dataset.xlsx",
index=False
)

print("\nDataset cleaned successfully.")
print("Cleaned file saved as Cleaned_Dataset.xlsx")
```
